```python
import librosa

audio, sr = librosa.load(
    "GuitarSet/audio-play/00_BN1-129-Eb_solo_mic.wav",
    sr=None
)

start = 3.240
end = 3.240 + 0.476

start_sample = int(start * sr)
end_sample = int(end * sr)

note_audio = audio[start_sample:end_sample]

print("Sample rate:", sr)
print("Start sample:", start_sample)
print("End sample:", end_sample)
print("Number of samples:", len(note_audio))
print("Duration:", len(note_audio) / sr)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

plt.plot(note_audio)

plt.title("Guitar Note Waveform")
plt.xlabel("Sample")
plt.ylabel("Amplitude")

plt.show()
```

```python
import numpy as np

fft = np.fft.rfft(note_audio)
frequencies = np.fft.rfftfreq(len(note_audio), 1 / sr)

magnitude = np.abs(fft)

peak_index = np.argmax(magnitude)

print("Peak frequency:", frequencies[peak_index], "Hz")
```
